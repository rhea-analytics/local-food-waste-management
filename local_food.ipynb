```python
# Dataabase Connection
conn = sqlite3.connect("food_waste_management.db")
cursor = conn.cursor()
```

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS providers(
        provider_ID INT PRIMARY KEY,
        Name VARCHAR(300),
        Type VARCHAR(100),
        Address TEXT,
        City VARCHAR(100),
        Contact VARCHAR(50)
    )
""")
conn.commit()
```

```python
for index, row in providers.iterrows():
    cursor.execute("""
        INSERT OR IGNORE INTO providers (Provider_ID, Name, Type, Address, City, Contact)
        VALUES (?, ?, ?, ?, ?, ?)
    """, (row['Provider_ID'], row['Name'], row['Type'], row['Address'], row['City'], row['Contact']))

conn.commit()
```

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS receivers(
        Receiver_ID INT PRIMARY KEY,
        Name VARCHAR(300),
        Type VARCHAR(100),
        City VARCHAR(100),
        Contact VARCHAR(50)
    )
""")
conn.commit()
```

```python
for index, row in receivers.iterrows():
    cursor.execute("""
        INSERT OR IGNORE INTO receivers (Receiver_ID, Name, Type, City, Contact)
        VALUES (?, ?, ?, ?, ?)
    """, (row['Receiver_ID'], row['Name'], row['Type'], row['City'], row['Contact']))

conn.commit()
```

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS food_listings(
        Food_ID INT PRIMARY KEY,
        Food_Name VARCHAR(300),
        Quantity VARCHAR(20),
        Expiry_Date VARCHAR(20),
        Provider_ID INTEGER,
        Provider_Type VARCHAR(50),
        Location VARCHAR(50),
        Food_Type VARCHAR(50),
        Meal_Type VARCHAR(50)
    )
""")
conn.commit()
```

```python
for index, row in food_listings.iterrows():
    cursor.execute("""
        INSERT OR IGNORE INTO food_listings (Food_ID, Food_Name, Quantity, Expiry_Date, Provider_ID, Provider_Type, Location, Food_Type, Meal_Type)
        VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
    """, (row['Food_ID'], row['Food_Name'], row['Quantity'], row['Expiry_Date'], row['Provider_ID'],row['Provider_Type'], row['Location'], row['Food_Type'], row['Meal_Type']))

conn.commit()
```

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS claims(
        Claim_ID INT PRIMARY KEY,
        Food_ID INT,
        Receiver_ID INT,
        Status VARCHAR(20),   
        Timestamp VARCHAR(20)
    )
""")
conn.commit()
```

```python
for index, row in claims.iterrows():
    cursor.execute("""
        INSERT OR IGNORE INTO claims (Claim_ID, Food_ID, Receiver_ID, Status, Timestamp)
        VALUES (?, ?, ?, ?, ?)
    """, (row['Claim_ID'], row['Food_ID'], row['Receiver_ID'], row['Status'], row['Timestamp']))

conn.commit()
```

```python
# 4. Which receivers have claimed the most food? 
query = """
SELECT receivers.Name,
SUM (food_listings.Quantity) AS total_claimed
FROM claims 
JOIN receivers ON claims.Receiver_ID = receivers.Receiver_ID
JOIN food_listings ON claims.Food_ID = food_listings.Food_ID
WHERE claims.Status = 'Completed'
GROUP BY receivers.Name
ORDER BY total_claimed DESC;
"""

top_receiver = pd.read_sql_query(query,conn)
print(top_receiver.head(1))
```

```python
# 9. Which provider has had the highest number of successful food claims? 
query = """
SELECT providers.Provider_ID,providers.Name,
COUNT (*) AS Successful_Claims
FROM claims
LEFT JOIN food_listings
ON claims.Food_ID = food_listings.Food_ID
LEFT JOIN providers
ON food_listings.Provider_ID=providers.Provider_ID
WHERE claims.Status = 'Completed'
GROUP BY food_listings.Provider_ID
ORDER BY Successful_Claims DESC
 LIMIT 1
"""
Successfully_Claimed = pd.read_sql_query(query,conn)
print(Successfully_Claimed)

```

```python
# 10. What percentage of food claims are completed vs. pending vs. canceled? 
query = """
SELECT Status, COUNT (*) * 100/(SELECT COUNT(*)FROM Claims)
AS Percentage
FROM claims
GROUP BY Status
"""
percentage_Claimed = pd.read_sql_query(query,conn)
print(percentage_Claimed)
```

```python
# 11. What is the average quantity of food claimed per receiver? 
query = """
SELECT receivers.Receiver_ID,receivers.Name,
Avg(food_listings.Quantity) AS Average_Quantity
FROM claims
JOIN receivers
ON claims.Receiver_ID = receivers.Receiver_ID 
JOIN food_listings
ON claims.Food_ID = food_listings.Food_ID
GROUP BY receivers.Receiver_ID,receivers.Name
"""

average_Claimed = pd.read_sql_query(query,conn)
print(average_Claimed)
```

```python
# 17. Which receiver type (e.g., NGO, individual) has made the most completed claims?
query = """
SELECT receivers.Type,
COUNT (*) AS high_claim
FROM receivers
LEFT JOIN claims
ON receivers.Receiver_ID = claims.Receiver_ID
WHERE claims.Status ='Completed'
GROUP BY receivers.Type
ORDER BY high_claim
LIMIT 1

"""
top = pd.read_sql_query(query,conn)
print(top)
```

```python
# 20. Are there any food listings that have never been claimed?
query = """
SELECT food_listings.Food_Name,
COUNT (claims.Food_ID) AS Never_Claimed
FROM food_listings
LEFT JOIN claims
ON food_listings.Food_ID = claims.Food_ID
WHERE claims.Food_ID IS NULL
GROUP BY food_listings.Food_Name
"""
never = pd.read_sql_query(query,conn)
print(never)
```
